```python
x=torch.tensor([[2.0 , 9.0],
   [4.0 , 7.0],
   [6.0 , 6.0],
   [8.0 , 5.0],
   [1.0 , 8.0]])
```

```python
y = torch.tensor([[0.4],[0.55],[0.7],[0.85],[0.3]])
```

```python
model = nn.Sequential(
    nn.Linear(2,4),
    nn.ReLU(),
    nn.Linear(4,1),
    nn.Sigmoid()
)
```

```python
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)
```

```python
print(f"\n{'Epoch':<8}{'Loss':<12}{'Whats Happening'}")
```

```python
for Epoch in range(201):
    predictions = model(x)
```

```python
loss = loss_fn(predictions,y)
```

```python
optimizer.zero_grad()
loss.backward()
```

```python
optimizer.step()
```

```python
if Epoch % 40 == 0:
    note = (
        "random weights , high loss" if Epoch == 0 else
        "Weight Adjusting" if Epoch > 120 else
        "Convering" if Epoch > 200 else
        "Trained"
    )
print(f"\n{'Epoch':<8}{loss.item():12.6f} {note}")
```
